• Calcular sobre el sueldo las deducciones por ley ISSS (3%), AFP(6.25%), ISR(Según tabla) e imprima el total deducciones y
sueldo neto. (20%

```python

# --- Funciones de cálculo ---

def calcular_isss(salario):
    """ISSS: 3% del salario, con tope de $1000.00 de base (osea máximo $30.00)"""
    base_cotizable = min(salario, 1000)
    return round(base_cotizable * 0.03, 2)


def calcular_afp(salario):
    """AFP: 7.25% del salario"""
    return round(salario * 0.0725, 2)


def calcular_isr(renta_gravable):
    """ISR mensual según tabla del Ministerio de Hacienda"""
    if renta_gravable <= 472.00:
        return 0.00
    elif renta_gravable <= 895.24:
        return round((renta_gravable - 472.00) * 0.10 + 17.67, 2)
    elif renta_gravable <= 2038.10:
        return round((renta_gravable - 895.24) * 0.20 + 60.00, 2)
    else:
        return round((renta_gravable - 2038.10) * 0.30 + 288.57, 2)


# Función para el sueldo neto
calcular_neto = lambda salario, deducciones: round(salario - deducciones, 2)


# --- Main ---

def main():
    print("=== Calculadora de Sueldo ===\n")

    nombre = input("Nombre del empleado: ")
    cargo = input("Cargo: ")
    salario_str = input("Sueldo bruto mensual: $")

    try:
        salario = float(salario_str)
        if salario < 0:
            raise ValueError
    except ValueError:
        print("Sueldo inválido. Ingrese un número positivo.")
        return

    isss = calcular_isss(salario)
    afp = calcular_afp(salario)
    renta_gravable = round(salario - isss - afp, 2)
    isr = calcular_isr(renta_gravable)

    total_deducciones = round(isss + afp + isr, 2)
    sueldo_neto = calcular_neto(salario, total_deducciones)

    print("\n--- Resultados ---\n")
    print(f"Empleado:          {nombre}")
    print(f"Cargo:             {cargo}")
    print(f"Sueldo Bruto:      ${salario:.2f}")
    print(f"ISSS (3%):         ${isss:.2f}")
    print(f"AFP (7.25%):       ${afp:.2f}")
    print(f"Renta Gravable:    ${renta_gravable:.2f}")
    print(f"ISR:               ${isr:.2f}")
    print(f"Total Deducciones: ${total_deducciones:.2f}")
    print(f"Sueldo Neto:       ${sueldo_neto:.2f}")


if __name__ == "__main__":
    main()
```

• Calcular el valor a pagar según el número de metros consumidos de agua potable (Ver tabla en la siguiente diapositiva).
(20%)

```python
def calcular_agua(metros):
    cuota_fija = 6.0

    if metros <= 18:
        total = cuota_fija

    elif metros <= 28:
        exceso = metros - 18
        total = cuota_fija + (exceso * 0.45)

    else:
        # De 19 a 28 (10 metros)
        exceso_primero = 10
        costo_primero = exceso_primero * 0.45

        # Más de 28
        exceso_segundo = metros - 28
        costo_segundo = exceso_segundo * 0.65

        total = cuota_fija + costo_primero + costo_segundo

    return total


# --- Programa principal con entrada del usuario ---
print("=" * 45)
print("CÁLCULO DE CONSUMO DE AGUA POTABLE")
print("=" * 45)

while True:
    try:
        metros = float(input("\nIngresa los metros cúbicos consumidos: "))

        if metros < 0:
            print("El consumo no puede ser negativo. Intenta de nuevo.")
            continue

        total = calcular_agua(metros)

        print("\n" + "-" * 45)
        print(f"Consumo: {metros:.2f} m³")
        print(f"Total a pagar: ${total:.2f}")
        print("-" * 45)

        # Preguntar si quiere calcular otro
        repetir = input("\n¿Quieres calcular otro consumo? (s/n): ").lower()
        if repetir != 's':
            print("\n¡chaito ahi te vez!")
            break

    except ValueError:
        print("Error: Ingresa un número válido.")
```

 • Desarrolle la tabla de meses acumulados a n meses. Si puede encontrar la fórmula matemática para evitar ciclos. (10%)

```python
n = 150  # número de meses a mostrar

print(f"{'MESES':<10}{'ACUMULACION'}")
for mes in range(1, n + 1):
    acumulado_anterior = mes * (mes - 1) // 2   # fórmula de Gauss hasta mes-1
    acumulado = mes * (mes + 1) // 2            # fórmula de Gauss hasta mes
    print(f"{mes:<10}{mes} + {acumulado_anterior} = {acumulado}")
```

• Realizar el sistema de gestión del tiempo y la demanda en las unidades de salud.

```python
from datetime import datetime

"""
- Prioridad I / Rojo:
    atención inmediata o de emergencia, el usuario debeser
    evaluado, estabilizado y recibir tratamiento por el personal de mayor capacidadresolutiva de la unidad de salud y luego ser referido al nivel decorrespondiente, en caso de ser

- Prioridad II /
    Amarillo: Atención en menos de 2 horas, debe explicar al usuario/aosuacompañante el tiempo
    promedio de espera, se debe dar prioridad a embarazadas, puérperas, neonatos, niños y niñas menores de un año, adultos/as mayores, personascon ENT descompensadas, personas con discapacidad y personas que demandan

- Prioridad III / Verde:
    Atención en 2 horas (máximo 4 horas), se le debe explicar al
    usuario/a los tiempos de espera establecidos para recibir la atención. En esta prioridadse incluyen todas las atenciones no contempladas en prioridad I ó II.

"""

# Motivos de consulta, agrupados por área/tipo de atención
motivos_consulta = {
    "consulta_general": {
        "Fiebre": "III",
        "Gripe o resfriado": "III",
        "Dolor de cabeza": "III",
        "Chequeo de rutina": "III",
        "Control de presión arterial": "III",
        "Vacunacion": "III"
    },
    "emergencia": {
        "Accidente de tránsito": "I",
        "Dificultad para respirar": "I",
        "Dolor de pecho": "I",
        "Hemorragia severa": "I",
        "Pérdida de conciencia": "I",
    },
    "pediatria": {
        "Vacunación": "III",
        "Fiebre en niños": "II",
        "Control de crecimiento": "III",
    },
    "ginecologia_obstetricia": {
        "Control prenatal": "II",
        "Dolor abdominal": "II",
        "Trabajo de parto": "I",
    },
    "odontologia": {
        "Dolor dental": "III",
        "Limpieza dental": "III",
        "Extraccion de diente": "III",
    },
    "traumatologia": {
        "Fractura": "II",
        "Esguince": "III",
        "Herida abierta": "II",
    },
    "retiros": {
        "Retiro de medicinas": "III",
        "Retiro de resultados de exámenes": "III",
        "Retiro de constancia o certificado médico": "III",
        "Retiro de insumos médicos": "III",
    }
}

# Motivos donde la edad puede subir el nivel de prioridad
motivos_sensibles_a_edad = {
    "Fiebre",
    "Gripe o resfriado",
    "Control de presión arterial",
    "Dolor de cabeza",
    "Fiebre en niños",
    "Fractura",
    "Herida abierta",
}

# Pacientes ya atendidos (historial del turno)
atendidos = []


# Niveles de prioridad/triaje
prioridades = {
    "I": [],    # Emergencia — riesgo vital inmediato, atención inmediata
    "II": [],   # Urgente — atención en minutos, no puede esperar mucho
    "III": [],  # Menos urgente — puede esperar en cola normal
}

# Menu principal de opciones para que consulta
def mostrarTipoConsulta():
    k = 0
    for i in motivos_consulta:
        k += 1
        print(f"{k}. {i.replace("_"," ")}")

# Guarda los datos en las prioridades
def guardarDatos(paciente, prioridad):
    # Agregar el paciente a la lista correspondiente según su prioridad
    prioridades[prioridad].append(paciente)
    print(f"Paciente {paciente["nombre"]} registrado con prioridad {prioridad}.")

# Capturar datos de paciente
def capturarDatosPaciente():

    embarazada = False

    nombre = input("Ingrese el nombre del paciente: ")
    edad = float(input("Ingrese la edad del paciente: "))
    print("1. Sexo masculino")
    print("2. Sexo memenino")

    respuesta = input("")
    if respuesta == "1":
        sexo = "Masculino"
        embarazada = False
    elif respuesta == "2":
        sexo = "Femenino"
        embarazada = False
        if 12 <= edad <= 50:  # rango más realista de edad fértil
            resp_embarazo = input("¿La paciente está embarazada? (s/n): ").strip().lower()
            embarazada = resp_embarazo == "s"   # <-- aquí estaba el bug

    print("")
    print("==== Motivo ====")
    mostrarTipoConsulta()
    print("")
    area_seleccionada = int(input("Seleccione el área de consulta: "))

    #Obtener el area selecionada por el usuario
    areaLista = list(motivos_consulta.keys())
    area = areaLista[area_seleccionada-1]

    print("")
    mostrarMotivosConsulta(area)
    print("")

    motivo_seleccionado = int(input("Seleccione el motivo de consulta: "))
    motivosLista = list(motivos_consulta[area].keys())
    motivo = motivosLista[motivo_seleccionado - 1]
    print(motivo)
    print("Ingrese un numero valido")

    # Antes de guardar, determinamos la prioridad del pacienye segun el motivo
    prioridad = determinarPrioridad(area, motivo, edad, embarazada)

    fecha = datetime.now()

    # Guardar los datos del paciente
    paciente = {
        "nombre": nombre,
        "edad": edad,
        "sexo": sexo,
        "embarazada": embarazada,
        "motivo": motivo,
        "prioridad": prioridad,
        "horaFecha": fecha
    }

    #Guardar el paciente
    guardarDatos(paciente, prioridad)

# Clasisificar la edad para la prioridad
def clasificar_edad(edad):
    if edad < (1 / 12): # menos de 1 mes
        return "neonato"
    elif edad < 2:      # 1 mes - 2 años
        return "lactante"
    elif edad < 12:     # 2 - 12 años
        return "nino"
    elif edad < 18:     # 12 - 18 años
        return "adolescente"
    elif edad < 65:     # 18 - 65 años
        return "adulto"
    else:               # 65+
        return "adulto_mayor"


# Determinar la prioridad necesaria segun el motivo y la edad
def determinarPrioridad(area, motivo, edad, embarazada=False):
    grupo = clasificar_edad(edad)
    grupos_riesgo_alto = ("neonato", "lactante")
    grupos_riesgo_medio = ("nino", "adulto_mayor")

    niveles = ["I", "II", "III"]

    # Caso especial: emergencia obstétrica / trabajo de parto siempre es I
    if embarazada and motivo == "Trabajo de parto":
        return "I"

    base = motivos_consulta.get(area, {}).get(motivo, "III")
    idx = niveles.index(base)

    if idx == 0:
        return "I"

    # Ajuste por edad (como ya lo tenías)
    if motivo in motivos_sensibles_a_edad:
        if grupo in grupos_riesgo_alto:
            idx = max(idx - 2, 0)
        elif grupo in grupos_riesgo_medio:
            idx = max(idx - 1, 0)

    # Ajuste por embarazo: nunca puede quedar en III si está embarazada
    if embarazada:
        idx = min(idx, niveles.index("II"))  # sube a II como mínimo

    return niveles[idx]

# Imprime los motivos para la consulta
def mostrarMotivosConsulta(area):
    k = 0
    print(f"==== Motivos de consulta para {area.replace("_"," ")} ====")
    for i in motivos_consulta[area]:
        k += 1
        print(f"{k}. {i}")


# Dar paso al siguiente turno para atenderlo
def siguienteTurno():
    for prioridad in ["I", "II", "III"]:
        if prioridades[prioridad]:
            paciente = prioridades[prioridad].pop(0)  # Sacar al primer paciente de la lista
            paciente["horaAtencion"] = datetime.now()  # Registrar cuándo se le atendió
            atendidos.append(paciente)                 # Guardarlo en el historial
            print(f"Atendiendo a {paciente['nombre']}: {paciente['motivo']}")
            return
    print("No hay pacientes en espera.")

# Imprmir los pacientes almeacenados por mayor prioridad
def mostrarPacientesPorPrioridad():
    for prioridad, pacientes_lista in prioridades.items():
        if pacientes_lista:
            for paciente in pacientes_lista:
                # Si es una embarazada muestra el campo
                if paciente["embarazada"] == True:
                        print(f"Nombre: {paciente['nombre']} \nEdad: {paciente['edad']} \nSexo: {paciente['sexo']}\nEmbarazada\nMotivo: {paciente['motivo']} \nPrioridad: {paciente['prioridad']}\nHora de llegada: {paciente['horaFecha']}\n")

                else:
                    print(f"Nombre: {paciente['nombre']} \nEdad: {paciente['edad']} \nSexo: {paciente['sexo']}\nMotivo: {paciente['motivo']} \nPrioridad: {paciente['prioridad']}\nHora de llegada: {paciente['horaFecha']}\n")

def mostrarAtendidos():
    if not atendidos:
        print("Todavía no se ha atendido a ningún paciente.")
        return

    print(f"==== Pacientes atendidos ({len(atendidos)}) ====")
    for paciente in atendidos:
        print(f"Nombre: {paciente['nombre']} \nEdad: {paciente['edad']} \nSexo: {paciente['sexo']}\nMotivo: {paciente['motivo']} \nPrioridad: {paciente['prioridad']}\nHora de llegada: {paciente['horaFecha']}\nHora de atención: {paciente['horaAtencion']}\n")

# Main donde funciona todo el programa
def main():
    print("===== Bienvenido a gestion de tiempo y demanda ======")
    while True:

        print("""
==== Opciones =====
1. Anotar paciente
2. Mostrar en espera por prioridad
3. siguiente turno
4. mostrar atendidos
Pulse cualquier otra tecla para salir
""")
        opcion = input("Seleccione una opción: ")
        if opcion == "1":
            print("")
            capturarDatosPaciente()
        elif opcion == "2":
            print("==== En espera ====")
            mostrarPacientesPorPrioridad()
        elif opcion == "3":
            print("==== Atendiento al siguiente ====")
            siguienteTurno()
        elif opcion == "4":
            print("==== Vitacora ====")
            mostrarAtendidos()
        else:
            print("Saliendo del sistema.")
            break

# Punto de inicio
main()
```
